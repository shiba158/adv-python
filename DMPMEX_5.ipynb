```python
import numpy as np
from itertools import combinations

# Our transactions
transactions = [
    ['Milk', 'Bread', 'Butter'],
    ['Bread', 'Butter'],
    ['Milk', 'Bread'],
    ['Milk', 'Butter'],
    ['Milk', 'Bread', 'Butter']
]

# Minimum support = 60%
min_support = 0.6

# Get all different items
items = ['Milk', 'Bread', 'Butter']

print("Items:", items)
print("\nFrequent Itemsets:")

# Try itemsets of size 1, 2, and 3
for size in range(1, len(items) + 1):

    # Create combinations
    for itemset in combinations(items, size):

        count = 0

        # Check every transaction
        for transaction in transactions:

            # If all items in the itemset are in the transaction
            if all(item in transaction for item in itemset):
                count += 1

        # Calculate support
        support = count / len(transactions)

        # Print if support is at least 60%
        if support >= min_support:
            print(itemset, "Support =", support * 100, "%")
```

```python
import pandas as pd
import numpy as np

# Step 1: Create transaction data
data = [
    ['Milk', 'Bread', 'Butter'],
    ['Bread', 'Butter'],
    ['Milk', 'Bread'],
    ['Milk', 'Butter'],
    ['Bread', 'Butter', 'Jam']
]

# Convert to DataFrame
df = pd.DataFrame(data)
print("Transactions:")
print(df)


# Step 2: Find support count of each item
items = np.unique([item for transaction in data for item in transaction])

support = {}

for item in items:
    count = sum(item in transaction for transaction in data)
    support[item] = count

print("\nItem Support:")
for item, count in support.items():
    print(item, ":", count)


# Step 3: Set minimum support
min_support = 2

frequent_items = {
    item: count
    for item, count in support.items()
    if count >= min_support
}

print("\nFrequent Items:")
print(frequent_items)


# Step 4: Find frequent pairs
from itertools import combinations

frequent_pairs = {}

for item1, item2 in combinations(frequent_items.keys(), 2):
    count = sum(
        item1 in transaction and item2 in transaction
        for transaction in data
    )

    if count >= min_support:
        frequent_pairs[(item1, item2)] = count


print("\nFrequent Pairs:")
for pair, count in frequent_pairs.items():
    print(pair, ":", count)


# Step 5: Find frequent triples
frequent_triples = {}

for items in combinations(frequent_items.keys(), 3):
    count = sum(
        all(item in transaction for item in items)
        for transaction in data
    )

    if count >= min_support:
        frequent_triples[items] = count


print("\nFrequent Triples:")
for triple, count in frequent_triples.items():
    print(triple, ":", count)
```

```python
import pandas as pd
import numpy as np
from itertools import combinations

# Transactions
data = [
    ['Milk', 'Bread'],
    ['Bread', 'Butter'],
    ['Milk', 'Bread', 'Butter'],
    ['Milk', 'Butter'],
    ['Bread', 'Butter']
]

# Total number of transactions
n = len(data)

# Minimum support and confidence
min_support = 0.4
min_confidence = 0.6

# Find all items
items = np.unique([item for t in data for item in t])

# Function to calculate support
def support(itemset):
    count = sum(
        set(itemset).issubset(set(t))
        for t in data
    )
    return count / n


# Generate frequent itemsets
frequent = {}

for size in range(1, len(items) + 1):
    for combo in combinations(items, size):
        s = support(combo)

        if s >= min_support:
            frequent[combo] = s

print("Frequent Itemsets:")
for itemset, s in frequent.items():
    print(itemset, "Support =", round(s, 2))


# Generate Association Rules
print("\nAssociation Rules:")

for itemset, itemset_support in frequent.items():

    if len(itemset) < 2:
        continue

    # Generate possible left-hand sides
    for i in range(1, len(itemset)):
        for left in combinations(itemset, i):

            right = tuple(
                item for item in itemset
                if item not in left
            )

            left_support = support(left)

            confidence = itemset_support / left_support

            if confidence >= min_confidence:
                print(
                    left,
                    "->",
                    right,
                    "Confidence =",
                    round(confidence, 2)
                )
```
